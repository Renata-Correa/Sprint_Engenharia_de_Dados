#🪞 Autoavaliação

A realização deste projeto foi **um grande desafio** para mim, principalmente por eu estar em transição de carreira para a área de dados, vindo de uma formação e atuação em outra área.

Construir um pipeline de dados completo, desde a ingestão até a camada analítica, exigiu dedicação extra, estudo contínuo e muita tentativa e erro, treinos e mão na massa com tecnologias que ainda estavam em processo de consolidação no meu aprendizado.

Ao longo do desenvolvimento, enfrentei dificuldades relacionadas à modelagem dos dados, ao uso do Databricks Free Edition, a criação de tabelas Delta e à aplicação correta do esquema medalhão. Apesar disso, cada obstáculo contribuiu diretamente para o fortalecimento da minha base técnica e para o aumento da minha autonomia na resolução de problemas.

**Mesmo com os desafios, o processo foi extremamente gratificante**.

Ver os dados evoluindo ao longo das camadas do pipeline, performando corretamente e, principalmente, conseguindo responder as perguntas de negócio que defini no início do projeto, trouxe uma sensação clara de conquista e validação do meu esforço. ✨

Esse projeto reforçou minha compreensão sobre a importância da Engenharia de Dados como fundamento para análises confiáveis e decisões estratégicas.

Além disso, confirmou minha motivação em **seguir me desenvolvendo na área de dados**, mostrando na prática que, mesmo vindo de outro contexto profissional, sou capaz de construir soluções estruturadas, funcionais e alinhadas as necessidades do negócio.

Considero este projeto um **marco importante na minha trajetória profissional**, tanto pelo aprendizado técnico quanto pela consolidação da minha transição para a área de dados.
